```python
# Zad. 14.1

import pandas as pd
import matplotlib.pyplot as plt
plt.style.use('default')  # reset stylu
plt.rcParams['figure.facecolor'] = 'white'
plt.rcParams['axes.facecolor'] = 'white'
import seaborn as sns
import numpy as np

fig, axes = plt.subplots(2, 3, figsize=(18, 10))

sns.histplot(df_meta['deltaT_label'], kde=False, ax=axes[0, 0], color='skyblue')
axes[0, 0].set_title('Rozkład temperatur (deltaT)')

sns.histplot(df_meta['pressure_corr_mpa'], kde=False, bins=100, ax=axes[0, 1], color='salmon')
axes[0, 1].set_title('Rozkład ciśnień (MPa)')

df_meta['point_type'].value_counts().plot(kind='pie', autopct='%1.1f%%', ax=axes[0, 2])
axes[0, 2].set_title('Udział typów punktów (point_type)')

df_meta['series_id'].value_counts().head(10).plot(kind='bar', ax=axes[1, 0])
axes[1, 0].set_title('Rozkład 10 najczęstszych serii')

counts = df_meta['is_complete_pair'].value_counts()
counts.plot(kind='bar', ax=axes[1, 1], color=['green', 'red'])
axes[1, 1].set_title('Kompletne vs niekompletne pary X/Y')
labels = [('Kompletne' if x == True else 'Niekompletne') for x in counts.index]
axes[1, 1].set_xticks(range(len(labels)))
axes[1, 1].set_xticklabels(labels, rotation=0)

fig.delaxes(axes[1, 2])
plt.tight_layout()
plt.savefig("Output_files/14_1.png")
plt.show()
```

```python
df_meta["pressure_num"] = (
    df_meta["pressure_raw_label"]
    .str.replace("MPa", "", regex=False)
    .astype(float)
)

middle = sorted(
    x for x in df_meta["pressure_num"].unique()
    if x not in [0, 0.01, 11]
)

x_order_num = [0] + middle + [0.01, 11]

x_order_labels = [f"{x:g} MPa" for x in x_order_num]

# Create categorical labels
df_meta["pressure_plot"] = pd.Categorical(
    df_meta["pressure_num"].map(lambda x: f"{x:g} MPa"),
    categories=x_order_labels,
    ordered=True
)

# OPTIONAL: same idea for temperature axis
temp_order = sorted(df_meta["T_plate_label"].unique())

df_meta["temp_plot"] = pd.Categorical(
    df_meta["T_plate_label"],
    categories=temp_order,
    ordered=True
)

# Count combinations
counts = (
    df_meta.groupby(["pressure_plot", "deltaT_label"])
    .size()
    .reset_index(name="count")
)

# Plot
plt.figure(figsize=(12, 6))

sns.scatterplot(
    data=counts,
    x="pressure_plot",
    y="deltaT_label",
    size="count",
    hue="count",
    palette="bright",
    sizes=(50, 1000),
    legend=True
)

plt.xticks(rotation=35)

plt.xlabel("Ciśnienie")
plt.ylabel("Temperatura")
plt.title("Liczba kombinacji ciśnienia i temperatury")

plt.grid(True)
plt.savefig("Output_files/kombinacja_pomiarów.png")
plt.show()
```

```python
# Zad. 14.2

df_plot = df_summary[df_summary["ref"].notna()].copy()

# 3 wykresy obok siebie
fig, axes = plt.subplots(1, 3, figsize=(21, 6))

# =========================
# 1. mu_X vs dT
# =========================
for e in sorted(df_plot['pressure'].unique()):
    df_p = df_plot[df_plot['pressure'] == e]

    sns.regplot(
        data=df_p,
        x='dT',
        y='mu_X',
        ax=axes[0],
        label=f'p={e}',
        scatter_kws={'s': 40}
    )

axes[0].set_title('mu_X vs Temperatura')
axes[0].set_xlabel('dT')
axes[0].set_ylabel('mu_X [GHz]')
axes[0].grid()
axes[0].legend(title='Ciśnienie', bbox_to_anchor=(1.02, 1),loc='upper left')


# =========================
# 2. mu_Y vs dT
# =========================
for e in sorted(df_plot['pressure'].unique()):
    df_p = df_plot[df_plot['pressure'] == e]

    sns.regplot(
        data=df_p,
        x='dT',
        y='mu_Y',
        ax=axes[1],
        label=f'p={e}',
        scatter_kws={'s': 40}
    )

axes[1].set_title('mu_Y vs Temperatura')
axes[1].set_xlabel('dT')
axes[1].set_ylabel('mu_Y [GHz]')
axes[1].grid()
axes[1].legend(title='Ciśnienie', bbox_to_anchor=(1.02, 1),loc='upper left')


# =========================
# 3. mu_X i mu_Y vs pressure
# =========================
df_p = df_plot[df_plot['dT'] == 0]

sns.regplot(
    data=df_p,
    x='pressure',
    y='mu_X',
    ax=axes[2],
    label='mu_X',
    scatter_kws={'s': 40}
)

sns.regplot(
    data=df_p,
    x='pressure',
    y='mu_Y',
    ax=axes[2],
    label='mu_Y',
    scatter_kws={'s': 40}
)

axes[2].set_title('Przesunięcie widmowe vs Ciśnienie (dT=0)')
axes[2].set_xlabel('Ciśnienie')
axes[2].set_ylabel('mu [GHz]')
axes[2].grid()
axes[2].legend()

plt.tight_layout()
plt.savefig("Output_files/14_2.png")
plt.show()
```

```python
# Zad. 14.3

features = ['dT', 'pressure', 'mu_X', 'mu_Y', 'diff_XY', 'mean_XY']
corr_matrix = df_plot[features].corr()

plt.figure(figsize=(10, 8))
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', fmt=".2f")
plt.title('Macierz korelacji cech bazowych z parametrami docelowymi')
plt.savefig("Output_files/14_3.png")
plt.show()


corr_p_diff = corr_matrix.loc['pressure', 'diff_XY']
corr_p_muX = corr_matrix.loc['pressure', 'mu_X']
corr_T_mean = corr_matrix.loc['dT', 'mean_XY']
corr_T_muX = corr_matrix.loc['dT', 'mu_X']

print(f"Korelacja p z diff_XY: {corr_p_diff:.4f} (vs mu_X: {corr_p_muX:.4f})")
print(f"Korelacja dT z mean_XY: {corr_T_mean:.4f} (vs mu_X: {corr_T_muX:.4f})")

print("Wnioski:")
if abs(corr_p_diff) > abs(corr_p_muX):
    print(" -diff_XY poprawia korelację z ciśnieniem względem pojedynczego kanału.")
else:
    print(" -diff_XY nie wykazuje silniejszej korelacji liniowej z ciśnieniem niż mu_X.")

if abs(corr_T_mean) > abs(corr_T_muX):
    print(" -mean_XY poprawia korelację z temperaturą względem pojedynczego kanału.")
else:
    print(" -mean_XY nie wykazuje silniejszej korelacji liniowej z temperaturą niż mu_X.")
```
